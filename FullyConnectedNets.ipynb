## Inline Question 1:
Did you notice anything about the comparative difficulty of training the three-layer network vs. training the five-layer network? In particular, based on your experience, which network seemed more sensitive to the initialization scale? Why do you think that is the case?

## Answer:

The **five-layer network** was noticeably more difficult to train and significantly more sensitive to the weight initialization scale than the three-layer network.

1. **Narrow Hyperparameter Basin for Initialization:**
   In the three-layer network, a broader range of `weight_scale` values (e.g., $10^{-2}$ to $10^{-1}$) led to rapid convergence and 100% training accuracy. In contrast, the five-layer network required precise tuning of `weight_scale` (around $4\times 10^{-2}$ to $5\times 10^{-2}$). Deviating slightly resulted either in a complete failure to learn (loss stuck around 2.3) or numerical instability (exploding loss / NaNs).

2. **Theoretical Mechanism — Repeated Matrix Multiplications:**
   At each layer $l$, the forward signal involves repeated linear transformations and non-linearities:
   $$h^{[l]} = \text{ReLU}\left(W^{[l]} h^{[l-1]} + b^{[l]}\right)$$
   
   - **Vanishing Activations & Gradients:** If the weight scale is slightly too small (e.g., $10^{-5}$ as in the default setup), each forward pass attenuates the magnitude of activations exponentially across the depth:
     $$\text{Var}(h^{[L]}) \propto \prod_{l=1}^L \text{Var}(W^{[l]})$$
     By the fifth layer, the pre-activations shrink toward zero. In the backward pass, backpropagated gradients scale by $W^T$ at each layer, causing gradients to vanish exponentially. As a result, early layers receive virtually zero weight updates.
   - **Exploding Activations & Gradients:** Conversely, if `weight_scale` is even slightly too large, activations grow exponentially through the layers, pushing values into extreme ranges, destabilizing ReLU activations, and causing gradient explosions.

Because the exponential factor is $L$ (depth), a deeper network ($L=5$) amplifies these compounding scaling effects much more drastically than a shallower network ($L=3$), making it far more sensitive to the choice of initialization scale in the absence of techniques like Batch Normalization or He (Kaiming) initialization.

## Inline Question 2:

AdaGrad, like Adam, is a per-parameter optimization method that uses the following update rule:

```
cache += dw**2
w += - learning_rate * dw / (np.sqrt(cache) + eps)
```

John notices that when he was training a network with AdaGrad that the updates became very small, and that his network was learning slowly. Using your knowledge of the AdaGrad update rule, why do you think the updates would become very small? Would Adam have the same issue?


## Answer:

1. **Why AdaGrad updates become very small:**
   In AdaGrad, the squared gradients are accumulated monotonically across all iterations without decay:
   $$\text{cache}_t = \text{cache}_{t-1} + (\nabla_w L)^2 = \sum_{\tau=1}^t (\nabla_w L_\tau)^2$$
   Because $(\nabla_w L)^2 \ge 0$, the `cache` value grows monotonically larger with every training step. Consequently, the effective per-parameter learning rate:
   $$\alpha_{\text{eff}} = \frac{\alpha}{\sqrt{\text{cache}_t} + \epsilon}$$
   shrinks continuously toward zero. Over prolonged training, the denominator becomes so large that parameter updates vanish ($\Delta w \to 0$), effectively freezing the network weights prematurely before reaching a good local minimum.

2. **Does Adam have the same issue?**
   **No, Adam does not suffer from this issue.**
   
   Unlike AdaGrad, Adam uses an exponentially decaying moving average (leaky integration) of past squared gradients parameterized by $\beta_2 \in (0, 1)$ (typically $0.999$):
   $$v_t = \beta_2 v_{t-1} + (1 - \beta_2) (\nabla_w L_t)^2$$
   Because the older gradients are exponentially discounted by $\beta_2$, the second-moment accumulator $v_t$ represents the *current local scale* of gradient magnitudes rather than their sum over all historical iterations. If gradients decrease or stabilize, $v_t$ scales down accordingly, preventing the denominator from growing unbounded and allowing the network to continue making substantial updates throughout late training stages.
